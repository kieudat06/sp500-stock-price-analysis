```python
import numpy as np
import pandas as pd

df = pd.read_csv('../data/raw/data_raw.csv')
df.head()
```

```python
print(f"Số dòng: {df.shape[0]}")
print(f"Số cột: {df.shape[1]}")
```

```python
df.info()
```

```python
df.describe().round(2)
```

```python
missing = pd.DataFrame({
    "Missing": df.isnull().sum(),
    "Percent (%)": (df.isnull().mean()*100).round(2)
})

missing
```

```python
duplicates = df.duplicated().sum()

print("Số dòng trùng lặp:", duplicates)
```

```python
print("Số mã cổ phiếu:", df["Ticker"].nunique())

df["Ticker"].value_counts().head(10)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Từ:", df["Date"].min())
print("Đến:", df["Date"].max())
```

```python
cols = ["Open","High","Low","Close","Adj Close","Volume"]

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    outlier = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]

    print(col, ":", len(outlier))
```

```python
df_clean = df.copy()
df_clean.to_csv("../data/processed/data_clean.csv",index=False)

print("Đã lưu:", df_clean.shape)
```
